```python
from lxml import objectify
path="C:\\Users\\Yyx_2\\Desktop\\code\\company_data.xml"
with open(path, 'r', encoding='utf-8') as f:
     parsed = objectify.parse(f)
     root = parsed.getroot()
```

```python
data=[]
skip_fields=[]
for elt in root.employees.employee:
    el_data={}
    for child in elt.getchildren():
        if child.tag in skip_fields:
            continue
        el_data[child.tag]=child.text
    data.append(el_data)
pd.DataFrame(data)
```
