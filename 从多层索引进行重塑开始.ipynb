```python
import pandas as pd
import numpy as np

data=pd.DataFrame(np.arange(6).reshape((2,3)),
     index=pd.Index(['Alla','Babo'],name='AB'),
     columns=pd.Index(['one','two','three'],
                name='number'))
data
```

```python
ki=data.stack()
ki
```

```python
kd=ki.unstack('AB')#kd=ki.unstack(0)效果相同
kd
```

```python
s1=pd.Series([0,1,2,3],index=['a','b','c','d'])
s2=pd.Series([4,5,6],index=['c','d','e'])
data1=pd.concat([s1,s2],keys=['A','B'])
data1
```

```python
ki1=data1.unstack()
ki1#缺失值的产生
```

```python
kd1=ki1.stack()
kd1#可逆的，缺失值的省略
```

```python
kd1_1=data1.unstack().stack(future_stack=True)#kd1_1=data1.unstack().stack(dropna=False)保留na，但是这个用法好像更新了
kd1_1
```

```python
di_0=pd.DataFrame({'left0':ki,'right0':ki-2},
               columns=pd.Index(['left0','right0'],name='side'))
di_0
```

```python
di_1=pd.DataFrame({'left':ki,'right':ki-2},
               columns=pd.Index(['left0','right0'],name='side'))
di_1#参考不对应的情况（be like:left0和left）
```

```python
di_0.unstack('AB')#堆叠后变为最低层级
```

```python
di_0.unstack('AB').stack('side')#指出堆叠的轴向（side）
```
